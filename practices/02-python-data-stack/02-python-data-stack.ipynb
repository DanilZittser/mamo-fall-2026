## Изменяемые и неизменяемые объекты

- **неизменяемые** (англ. *immutable*): `int`, `float`, `bool`, `str`, `tuple`, `None` — любая «модификация» создаёт новый объект;
- **изменяемые** (англ. *mutable*): `list`, `dict`, `set` — меняются на месте.

Переменная в Python — это имя, которое ссылается на объект. Функция `id()` возвращает идентификатор объекта. В стандартной реализации Python (CPython) заранее создаются объекты целых чисел от −5 до 256. Кэширование небольших целых чисел — это оптимизация CPython, позволяющая экономить память и время на создании часто используемых объектов.

```python
x = 13  # связываем переменную x с объектом целого числа 13

id(13), id(x), id(13) == id(x)
```

В то же время для больших чисел:

```python
x = int('1000')
y = int('1000')
# значения равны, но это разные объекты
x == y, x is y
```

```python
text = 'abc'
text_id = id(text)
text += 'd'

text, id(text) == text_id
```

```python
items = [1, 2, 3]
items_id = id(items)
items += [4]

items, id(items) == items_id
```

Два имени могут ссылаться на один и тот же изменяемый объект — изменение через одно имя видно через другое:

```python
first = [1, 2, 3]
second = first
second.append(4)

first, first is second
```

```python
third = first.copy()  # Явная копия — новый объект
third.append(5)

first, third
```

## Функции

### Базовый синтаксис

Функция объявляется через `def`, возвращает значение через `return`. Без `return` функция возвращает `None`.

```python
def circle_area(radius):
    """Вычисляет площадь круга."""
    return 3.14159 * radius ** 2


circle_area(2)
```

### Позиционные и именованные аргументы, значения по умолчанию

### Ловушка: изменяемое значение по умолчанию

Значение по умолчанию вычисляется **один раз** — при объявлении функции. Если это список, все вызовы делят один и тот же объект.

```python
def append_item(item, items=[]):
    items.append(item)
    return items
```

```python
append_item(1)
```

```python
append_item(2)
```

Правильный способ — `None` по умолчанию и создание объекта внутри функции:

```python
def append_item(item, items=None):
    if items is None:
        items = []
    items.append(item)
    return items


append_item(1), append_item(2)
```

### Функция как аргумент: `sorted(key=...)` и `lambda`

Функции в Python — обычные объекты: их можно передавать в другие функции. `lambda` — короткая анонимная функция из одного выражения.

```python
def get_second_item(x):
    """Функция возвращает второй (нумерация с нуля) элемент коллекции."""
    return x[1]
```

```python
students = [('Вера', 22), ('Борис', 18), ('Анна', 20)]

sorted(students)
```

```python
sorted(students, key=get_second_item)  # выполняем сортировку по возрасту
```

То же самое можно выполнить при помощи `lambda`-функции:

```python
sorted(students, key=lambda x: x[1])
```

```python
sorted(['banana', 'Apple', 'cherry'], key=lambda x: x.lower())
```

```python
max(students, key=lambda student: student[1])
```

## Классы

Класс объединяет данные (атрибуты) и поведение (методы). Метод `__init__` вызывается при создании объекта, `self` — ссылка на сам объект.

```python
class Point:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def distance_to(self, other):
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5


start = Point(0, 0)
end = Point(3, 4)

start.distance_to(end)
```

```python
start
```

### [`@dataclass`](https://docs.python.org/3/library/dataclasses.html)

Для классов, которые в основном хранят данные, `@dataclass` сам генерирует `__init__`, `__repr__` и `__eq__` по аннотациям полей.

```python
from dataclasses import dataclass


@dataclass
class Point:
    x: float
    y: float

    def distance_to(self, other):
        return ((self.x - other.x) ** 2 + (self.y - other.y) ** 2) ** 0.5


start = Point(0, 0)
end = Point(3, 4)

start, start.distance_to(end), Point(1, 2) == Point(1, 2)
```

## Векторизация

Посчитаем сумму квадратов миллиона чисел — циклом Python и векторной операцией NumPy.

```python
values = rng.random(1_000_000)
values_list = list(values)
```

```python
%%timeit -n 3 -r 3
total = 0.0
for value in values_list:
    total += value ** 2
```

```python
%%timeit -n 3 -r 3
total = (values ** 2).sum()
```

Векторная операция быстрее на два порядка. Правило: **если есть векторная операция — цикл не нужен**.

Типичный пример — центрирование данных: вычитаем из каждого столбца его среднее.

# Задание: погода

Ниже генерируется таблица ежедневных наблюдений погоды в нескольких городах за 2025 год и сохраняется в `data/weather.csv`. Выполните задания по порядку.

```python
def make_weather_csv(path, seed=0):
    """Генерирует синтетические наблюдения погоды и сохраняет их в CSV."""
    rng = np.random.default_rng(seed)
    dates = pd.date_range('2025-01-01', '2025-12-31', freq='D')
    base_temperatures = {'Тюмень': 2.0, 'Омск': 1.5, 'Томск': 0.5, 'Сочи': 15.0}
    frames = []
    for city, base_temperature in base_temperatures.items():
        season = -15 * np.cos(2 * np.pi * (dates.dayofyear.to_numpy() - 15) / 365)  # Минимум в середине января
        frames.append(pd.DataFrame({
            'city': city,
            'date': dates,
            'temperature': (base_temperature + season + rng.normal(0, 4, len(dates))).round(1),
            'precipitation': rng.exponential(1.5, len(dates)).round(1) * rng.integers(0, 2, len(dates)),
        }))
    pd.concat(frames, ignore_index=True).to_csv(path, index=False)


make_weather_csv(data_dir / 'weather.csv')
```

```python
weather['month'] = weather['date'].dt.month
by_city = weather.groupby('city').agg(
    mean_temperature=('temperature', 'mean'),
    total_precipitation=('precipitation', 'sum'),
)
by_city
```

```python
def mean_temperature_loop(weather, city):
    """Вычисляет среднюю температуру города циклом по строкам."""
    total = 0.0
    count = 0
    for _, row in weather.iterrows():
        if row['city'] == city:
            total += row['temperature']
            count += 1
    return total / count


def mean_temperature_vectorized(weather, city):
    """Вычисляет среднюю температуру города векторной операцией."""
    return weather.loc[weather['city'] == city, 'temperature'].mean()


mean_temperature_loop(weather, 'Тюмень'), mean_temperature_vectorized(weather, 'Тюмень')
```

```python
%timeit -n 3 -r 3 mean_temperature_loop(weather, 'Тюмень')
```

```python
%timeit -n 3 -r 3 mean_temperature_vectorized(weather, 'Тюмень')
```
